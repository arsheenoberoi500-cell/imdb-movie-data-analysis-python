```python
import pandas as pd
import numpy as np

# Define a function to clean and convert gross income to a numeric type
def clean_and_convert_gross_income(income_str):
    if pd.isna(income_str) or not isinstance(income_str, str):
        return np.nan # Return NaN for missing or non-string values

    cleaned_str = income_str.replace('$', '').replace(',', '').strip()

    if 'M' in cleaned_str.upper():
        cleaned_str = cleaned_str.upper().replace('M', '')
        try:
            return float(cleaned_str) * 1_000_000
        except ValueError:
            return np.nan
    elif 'K' in cleaned_str.upper():
        cleaned_str = cleaned_str.upper().replace('K', '')
        try:
            return float(cleaned_str) * 1_000
        except ValueError:
            return np.nan
    else:
        try:
            return float(cleaned_str)
        except ValueError:
            return np.nan

# Apply the cleaning function to the 'gross_income' column in df2
df2['gross_income'] = df2['gross_income'].apply(clean_and_convert_gross_income)

# Now calculate the mean gross income per genre
df2.groupby('genre')['gross_income'].mean().sort_values(ascending=False).head()
```
